```python
from langchain_core.prompts import ChatPromptTemplate

prompts=ChatPromptTemplate.from_messages(
    [
    {"role":"system","content":"You are a translator from English to {language}"},
    {"role":"user","content":"{query}"},
    ]
)


final_prompts=prompts.format_messages(language="Hidni", query="I love Programming!")
res=llm.invoke(final_prompts)
print(res.content)
```
